# 04. Time-Series Demand & Revenue Forecasting

This notebook covers:
1. Daily revenue aggregation from transaction history.
2. Time-series decomposition (trend, seasonality, noise).
3. Time-series revenue forecasting (Prophet / Holt-Winters).
4. Plotting 30-day projected revenue trajectory with 95% confidence intervals.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

forecast_df = pd.read_csv("../data/processed/forecasts.csv")
forecast_df['Date'] = pd.to_datetime(forecast_df['Date'])
forecast_df.tail()
```

## Forecast vs Actual Revenue Visualization

```python
plt.figure(figsize=(14, 6))

hist_data = forecast_df[forecast_df['DataType'] == 'Historical']
fc_data = forecast_df[forecast_df['DataType'] == 'Forecast']

plt.plot(hist_data['Date'], hist_data['ActualRevenue'], label='Historical Daily Revenue', color='#1f77b4', lw=1.5)
plt.plot(fc_data['Date'], fc_data['PredictedRevenue'], label='30-Day Forecasted Revenue', color='#ff7f0e', lw=2, linestyle='--')
plt.fill_between(fc_data['Date'], fc_data['LowerBound'], fc_data['UpperBound'], color='#ff7f0e', alpha=0.25, label='95% Confidence Bounds')

plt.title("E-Commerce Daily Revenue: Historical Trend vs 30-Day Forecast", fontsize=14)
plt.xlabel("Date")
plt.ylabel("Revenue ($)")
plt.legend(loc="upper left")
plt.tight_layout()
plt.show()
```
