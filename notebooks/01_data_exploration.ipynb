# 01. Exploratory Data Analysis & Data Sanitation

This notebook covers:
1. Loading raw e-commerce transaction data.
2. Summary statistics and data quality checks.
3. Handling duplicate records, null values, and invalid prices/quantities.
4. Revenue calculation and sales distributions across countries, categories, and payment methods.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import sys

# Set plot style
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

# Load cleaned transactions
data_path = Path("../data/processed/cleaned_transactions.csv")
df = pd.read_csv(data_path)
df['TransactionDate'] = pd.to_datetime(df['TransactionDate'])
df.head()
```

## Summary Statistics

```python
print(f"Total Clean Transactions: {len(df):,}")
print(f"Total Unique Customers:    {df['CustomerID'].nunique():,}")
print(f"Total Gross Revenue:       ${df['Revenue'].sum():,.2f}")
print(f"Average Order Spend:       ${df['Revenue'].mean():,.2f}")
df.describe()
```

## Category & Geographic Revenue Visualizations

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

cat_rev = df.groupby('ProductCategory')['Revenue'].sum().sort_values(ascending=False)
sns.barplot(x=cat_rev.values, y=cat_rev.index, ax=axes[0], palette="viridis")
axes[0].set_title("Revenue by Product Category", fontsize=14)
axes[0].set_xlabel("Gross Revenue ($)")

country_rev = df.groupby('Country')['Revenue'].sum().sort_values(ascending=False)
sns.barplot(x=country_rev.values, y=country_rev.index, ax=axes[1], palette="magma")
axes[1].set_title("Revenue by Country", fontsize=14)
axes[1].set_xlabel("Gross Revenue ($)")

plt.tight_layout()
plt.show()
```
