# 03. Customer Churn Prediction (Supervised ML)

This notebook covers:
1. Defining Churn label based on 60-day inactivity threshold.
2. Train/Test split & Feature selection.
3. Training RandomForest Classifier model.
4. Evaluating Accuracy, Precision, Recall, F1-Score, and ROC-AUC curve.
5. Feature Importance analysis and Churn Risk categorization.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, roc_curve, auc, confusion_matrix

sns.set_theme(style="whitegrid")

churn_df = pd.read_csv("../data/processed/churn_predictions.csv")
churn_df.head()
```

## Churn Risk Level Visualizations

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.histplot(data=churn_df, x='ChurnProbability', hue='RiskCategory', kde=True, ax=axes[0], palette='Set1')
axes[0].set_title("Churn Probability Distribution by Risk Category", fontsize=14)
axes[0].set_xlabel("Predicted Churn Probability")

risk_rev = churn_df.groupby('RiskCategory')['Monetary'].sum().reindex(['High Risk', 'Medium Risk', 'Low Risk'])
sns.barplot(x=risk_rev.index, y=risk_rev.values, ax=axes[1], palette='Reds_r')
axes[1].set_title("Lifetime Revenue Exposure at Risk ($)", fontsize=14)
axes[1].set_ylabel("Total Monetary Spend ($)")

plt.tight_layout()
plt.show()
```
