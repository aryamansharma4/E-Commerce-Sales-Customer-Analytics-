# 02. Customer Segmentation (RFM + K-Means Clustering)

This notebook covers:
1. RFM (Recency, Frequency, Monetary) feature extraction.
2. Feature scaling using `StandardScaler`.
3. Elbow Method & Silhouette Analysis to evaluate optimal number of clusters ($k$).
4. Fitting K-Means clustering model.
5. Profiling cluster centroids and mapping business-oriented segment labels.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from pathlib import Path

sns.set_theme(style="whitegrid")

# Load engineered customer features
cust_df = pd.read_csv("../data/processed/customer_features.csv")
cust_df.head()
```

## RFM Feature Distributions & Scaling

```python
rfm_cols = ['Recency', 'Frequency', 'Monetary']
scaler = StandardScaler()
X_scaled = scaler.fit_transform(cust_df[rfm_cols])

# Elbow Method
inertia = []
    K_range = range(1, 8)
for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_scaled)
    inertia.append(km.inertia_)

plt.figure(figsize=(8, 4))
plt.plot(K_range, inertia, 'bo-')
plt.title("Elbow Method For Optimal k", fontsize=14)
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.show()
```

## Segment Analysis & Visualization

```python
seg_df = pd.read_csv("../data/processed/customer_segments.csv")

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
sns.scatterplot(
    data=seg_df, x='Recency', y='Monetary', hue='CustomerSegment',
    palette='Set2', alpha=0.8, ax=axes[0]
)
axes[0].set_title("Customer Segments: Recency vs Monetary Spend", fontsize=14)

seg_counts = seg_df['CustomerSegment'].value_counts()
axes[1].pie(seg_counts, labels=seg_counts.index, autopct='%1.1f%%', colors=sns.color_palette('Set2'))
axes[1].set_title("Segment Share Distribution", fontsize=14)

plt.tight_layout()
plt.show()
```
